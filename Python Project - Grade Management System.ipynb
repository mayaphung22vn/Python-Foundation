```python
from tkinter import * 
from tkinter import messagebox

window = Tk()

window.title("GPA management system")

list_marks=[]

def add_GPA():
    new_marks={
        "Name" : name.get(),
        "Semester1": float(Semester1.get()),
        "Semester2": float(Semester2.get()),
        }
    list_marks.append(new_marks)
    print(list_marks)

    messagebox.showinfo("Add marks", "Add marks successfully")

def search_GPA():
    search_value=Search.get()
    for i in range(len(list_marks)):
        marks=list_marks[i]
        if marks['Name']==search_value:
            messagebox.showinfo("Search result", f"FULL NAME : {marks['Name']}, GPA SEMESTER 1 : {marks['Semester1']} , GPA SEMESTER 2: {marks['Semester2']}")
        

def delete_GPA():
    delete_name=name.get()
    for i in range(len(list_marks)):
        marks=list_marks[i]
        if marks['Name']==delete_name:
            del list_marks[i]
            messagebox.showinfo("Delete Marks","Delete marks successfully")
            break 

def update_GPA():
    search_value=name.get()
    new_SEM1=float(Semester1.get())
    new_SEM2=float(Semester2.get())
    for i in range(len(list_marks)):
        marks=list_marks[i]
        if marks['Name'] == search_value:
            marks['Semester1']==new_SEM1
            marks['Semester2']==new_SEM2
            messagebox.showinfo("Update Marks", "Update marks successfully")

def reset():
    name.delete(0,END)
    Semester1.delete(0,END)
    Semester2.delete(0,END)
    Search.delete(0,END)

def check_GPA():
        Sem1=float(Semester1.get())
        Sem2=float(Semester2.get())
        average= (Sem1 + Sem2) / 2 
        if average >=3:
            messagebox.showinfo("Passing check", "Student passes exam")
        else:  
            messagebox.showinfo("Passing check", "Student fails exam")



name = Entry(window)
Semester1= Entry(window)
Semester2 = Entry(window)
Search= Entry(window)

name_lbl = Label(window,text="FULL NAME:")
Semester1_lbl= Label(window,text="SEMESTER 1:")
Semester2_lbl= Label(window,text="SEMESTER 2:")
 
add_btn= Button(window,text="Add", command=add_GPA)
delete_btn= Button(window,text="Delete",command=delete_GPA)
reset_btn= Button(window,text="Reset",command=reset)
search_btn= Button(window,text="Search",command=search_GPA)
update_btn= Button(window,text="Update",command=update_GPA)
check_btn= Button(window, text="Check", command=check_GPA)


name_lbl.grid(row=0,column=0)
name.grid(row=0,column=1)
Semester1_lbl.grid(row=1,column=0)
Semester1.grid(row=1,column=1)
Semester2_lbl.grid(row=2,column=0)
Semester2.grid(row=2,column=1)
add_btn.grid(row=3,column=0)
Search.grid(row=4,column=0)
search_btn.grid(row=4,column=1)
delete_btn.grid(row=3,column=1)
update_btn.grid(row=3,column=2)
reset_btn.grid(row=4,column=2)
check_btn.grid(row=3,column=3)

window.mainloop()
```
